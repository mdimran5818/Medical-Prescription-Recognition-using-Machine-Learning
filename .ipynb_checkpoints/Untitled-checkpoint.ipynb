```python
import os
import cv2
import numpy as np
from tensorflow.keras.models import load_model
from PIL import Image
import pytesseract
from transformers import BertTokenizerFast, BertForMaskedLM

# Define a function to preprocess the image for the classification model
def preprocess_image_for_classification(image_path, target_size=(128, 32)):
    img = cv2.imread(image_path)
    img = cv2.resize(img, target_size)
    img = img.astype('float32') / 255.0  # Normalize
    img = np.expand_dims(img, axis=0)  # Add batch dimension
    return img

# Define a function to preprocess the image for OCR
def preprocess_image_for_ocr(image_path):
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    img = cv2.resize(img, (128, 32))
    _, img = cv2.threshold(img, 150, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    return img

# Define a function to extract text using OCR
def extract_text_with_ocr(image_path):
    preprocessed_image = preprocess_image_for_ocr(image_path)
    img = Image.fromarray(preprocessed_image)
    extracted_text = pytesseract.image_to_string(img)
    return extracted_text

# Test classification model
def test_classification_model(model_path, image_path, medicine_names):
    model = load_model(model_path)
    img = preprocess_image_for_classification(image_path)
    predictions = model.predict(img)
    predicted_class = np.argmax(predictions)
    print(f"Predicted Medicine Name: {medicine_names[predicted_class]}")

# Test OCR model
def test_ocr_model(tokenizer_path, model_path, image_path):
    tokenizer = BertTokenizerFast.from_pretrained(tokenizer_path)
    model = BertForMaskedLM.from_pretrained(model_path)
    text = extract_text_with_ocr(image_path)
    print(f"Extracted Text: {text}")

    # Tokenize the extracted text
    inputs = tokenizer(text, return_tensors='pt', truncation=True, padding="max_length", max_length=128)
    with torch.no_grad():
        outputs = model(**inputs)
    
    predicted_ids = outputs.logits.argmax(dim=-1)
    decoded_text = tokenizer.decode(predicted_ids[0], skip_special_tokens=True)
    print(f"Decoded Text: {decoded_text}")

# Paths to your model and test image
classification_model_path = 'prescription_classification_model_CNRR.keras'
ocr_tokenizer_path = 'output_tokenizer'
ocr_model_path = 'output_model'
test_image_path = './Prescription dataset/Testing/testing_words/sample_test.png'  # Update with your test image path

# List of medicine names (make sure this matches your training set)
medicine_names = ['Aceta', 'Paracetamol', 'Ibuprofen', 'Baclon', ...]  # Complete the list

# Test the classification model
test_classification_model(classification_model_path, test_image_path, medicine_names)

# Test the OCR model
test_ocr_model(ocr_tokenizer_path, ocr_model_path, test_image_path)
```
